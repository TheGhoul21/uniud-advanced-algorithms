# String Matching — Interactive Companion

Interactive visualizations for Chapter 1 of the Advanced Algorithms notes.  
Use the sliders and buttons to step through each algorithm on your own inputs.

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import ipywidgets as widgets
from IPython.display import display, clear_output
import numpy as np

plt.rcParams.update({
    "font.family": "monospace",
    "font.size": 13,
    "figure.facecolor": "#fafafa",
})

COLORS = {
    "match": "#4CAF50",
    "mismatch": "#F44336",
    "current": "#FFC107",
    "skip": "#90CAF9",
    "default": "#E0E0E0",
    "pattern": "#BBDEFB",
    "border": "#7E57C2",
}


def draw_string_row(ax, y, string, label="", highlights=None, offset=0):
    """Draw a row of character boxes at vertical position y."""
    highlights = highlights or {}
    for i, ch in enumerate(string):
        color = highlights.get(i, COLORS["default"])
        rect = mpatches.FancyBboxPatch(
            (i + offset, y), 0.9, 0.9,
            boxstyle="round,pad=0.05",
            facecolor=color, edgecolor="#555", linewidth=1.2,
        )
        ax.add_patch(rect)
        ax.text(i + offset + 0.45, y + 0.45, ch,
                ha="center", va="center", fontsize=14, fontweight="bold")
    if label:
        ax.text(offset - 0.3, y + 0.45, label,
                ha="right", va="center", fontsize=12, color="#555")


print("Helpers loaded.")
```

---
## 1. Naive String Matching

Slide through every alignment of `P` against `T` and see which characters match.

```python
def naive_trace(T, P):
    """Return list of (shift, comparisons) where each comparison is (j, matched)."""
    steps = []
    for s in range(len(T) - len(P) + 1):
        comps = []
        for j in range(len(P)):
            matched = T[s + j] == P[j]
            comps.append((j, matched))
            if not matched:
                break
        steps.append((s, comps))
    return steps


def draw_naive(T, P, step_idx):
    steps = naive_trace(T, P)
    s, comps = steps[step_idx]
    last_j, last_matched = comps[-1]

    fig, ax = plt.subplots(figsize=(max(len(T) * 0.85, 8), 2.8))
    ax.set_xlim(-1.5, len(T) + 0.5)
    ax.set_ylim(-0.5, 2.5)
    ax.set_aspect("equal")
    ax.axis("off")

    # text row
    t_hi = {}
    for j, matched in comps:
        t_hi[s + j] = COLORS["match"] if matched else COLORS["mismatch"]
    draw_string_row(ax, 1.2, T, label="T", highlights=t_hi)

    # pattern row
    p_hi = {}
    for j, matched in comps:
        p_hi[j] = COLORS["match"] if matched else COLORS["mismatch"]
    draw_string_row(ax, 0, P, label="P", highlights=p_hi, offset=s)

    status = "MATCH FOUND!" if last_matched and last_j == len(P) - 1 else f"mismatch at j={last_j}" if not last_matched else "..."
    ax.set_title(f"shift s={s}  |  {status}", fontsize=13, pad=10)
    plt.tight_layout()
    plt.show()


T_input = widgets.Text(value="abcaabcabcaab", description="T:", layout=widgets.Layout(width="400px"))
P_input = widgets.Text(value="abcab", description="P:", layout=widgets.Layout(width="400px"))
step_slider = widgets.IntSlider(value=0, min=0, max=0, description="Step:")
out = widgets.Output()


def refresh(_=None):
    T, P = T_input.value, P_input.value
    if not T or not P or len(P) > len(T):
        return
    step_slider.max = len(T) - len(P)
    step_slider.value = min(step_slider.value, step_slider.max)
    with out:
        clear_output(wait=True)
        draw_naive(T, P, step_slider.value)


T_input.observe(refresh, "value")
P_input.observe(refresh, "value")
step_slider.observe(refresh, "value")

display(T_input, P_input, step_slider, out)
refresh()
```

## 2. KMP — Failure Function (sp values) & Search

Visualize the `sp` (border) array computation and then step through the KMP search showing how mismatches use the failure function to skip alignments.

```python
def compute_sp(P):
    """Compute sp[j] = length of longest proper border of P[0..j]."""
    m = len(P)
    sp = [0] * m
    k = 0
    for j in range(1, m):
        while k > 0 and P[k] != P[j]:
            k = sp[k - 1]
        if P[k] == P[j]:
            k += 1
        sp[j] = k
    return sp


def draw_sp_table(P):
    """Draw the pattern with its sp (border) array underneath."""
    sp = compute_sp(P)
    m = len(P)

    fig, ax = plt.subplots(figsize=(max(m * 0.85, 6), 3.2))
    ax.set_xlim(-1.5, m + 0.5)
    ax.set_ylim(-1.5, 2.5)
    ax.set_aspect("equal")
    ax.axis("off")

    # index row
    for i in range(m):
        ax.text(i + 0.45, 2.1, str(i), ha="center", va="center",
                fontsize=10, color="#999")

    # pattern row
    draw_string_row(ax, 1.0, P, label="P")

    # sp row
    sp_hi = {}
    for i in range(m):
        if sp[i] > 0:
            sp_hi[i] = COLORS["border"]
    for i in range(m):
        color = COLORS["border"] if sp[i] > 0 else "#F5F5F5"
        rect = mpatches.FancyBboxPatch(
            (i, -0.2), 0.9, 0.9,
            boxstyle="round,pad=0.05",
            facecolor=color, edgecolor="#555", linewidth=1.0,
        )
        ax.add_patch(rect)
        ax.text(i + 0.45, 0.25, str(sp[i]),
                ha="center", va="center", fontsize=14, fontweight="bold")
    ax.text(-0.3, 0.25, "sp", ha="right", va="center", fontsize=12, color="#555")

    # highlight border on last position
    if sp[-1] > 0:
        # prefix
        for j in range(sp[-1]):
            rect = mpatches.FancyBboxPatch(
                (j, 1.0), 0.9, 0.9,
                boxstyle="round,pad=0.05",
                facecolor=COLORS["border"], edgecolor="#555", linewidth=1.5,
            )
            ax.add_patch(rect)
            ax.text(j + 0.45, 1.45, P[j],
                    ha="center", va="center", fontsize=14, fontweight="bold")
        # suffix
        start = m - sp[-1]
        for j in range(sp[-1]):
            rect = mpatches.FancyBboxPatch(
                (start + j, 1.0), 0.9, 0.9,
                boxstyle="round,pad=0.05",
                facecolor=COLORS["border"], edgecolor="#555", linewidth=1.5,
            )
            ax.add_patch(rect)
            ax.text(start + j + 0.45, 1.45, P[start + j],
                    ha="center", va="center", fontsize=14, fontweight="bold")

    ax.set_title(f"sp (failure function) for P = \"{P}\"", fontsize=13, pad=10)
    plt.tight_layout()
    plt.show()


P_sp_input = widgets.Text(value="abcabcab", description="P:", layout=widgets.Layout(width="400px"))
out_sp = widgets.Output()


def refresh_sp(_=None):
    with out_sp:
        clear_output(wait=True)
        if P_sp_input.value:
            draw_sp_table(P_sp_input.value)


P_sp_input.observe(refresh_sp, "value")
display(P_sp_input, out_sp)
refresh_sp()
```

### KMP Search — Step by Step

Watch how KMP uses the failure function to avoid re-scanning characters. The arrow shows where `j` falls back to on a mismatch.
